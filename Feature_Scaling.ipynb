```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd
```

```python
df = pd.read_csv('data/data_cleaned.csv')
```

```python
df.head()
```

```python
numerical_features = ['Height (cm)', 'Width (cm)', 'Depth (cm)', 'Avg Daily Demand']
X = df[numerical_features]
```

```python
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
```

```python
df_scaled = df.copy()
df_scaled[numerical_features] = X_scaled
```

```python
df_scaled.head()
```

```python
df_scaled.to_csv('data/feature_scaled.csv', index=False)
```

**Label Encoding**

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()


df_scaled['Product_Label'] = le.fit_transform(df_scaled['Product Category'])
```

```python
df_scaled.head()
```

```python
from sklearn.preprocessing import OneHotEncoder

# Create OneHotEncoder object
encoder = OneHotEncoder(sparse_output=False)

# Apply One-Hot Encoding
encoded_data = encoder.fit_transform(df_scaled[['Product Category']])

# Create DataFrame for encoded columns
encoded_df = pd.DataFrame(
    encoded_data,
    columns=encoder.get_feature_names_out(['Product Category'])
)

# Concatenate with original dataframe
df_scaled = pd.concat([df_scaled, encoded_df], axis=1)

# Drop original categorical column
df_scaled.drop('Product Category', axis=1, inplace=True)

# Display result
print(df_scaled.head())
```

```python
df_scaled.to_csv('data/feature_scaled.csv', index=False)
```
